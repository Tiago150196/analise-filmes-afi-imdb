```python
def tabela_frequencia(coluna):

    frequencia = dados[coluna].value_counts()

    percentual = dados[coluna].value_counts(normalize=True) * 100

    tabela = pd.DataFrame({
        'Frequência': frequencia,
        'Porcentagem (%)': percentual
    })

    return tabela
```

```python
tabela_frequencia('Sexo')
```

Análise cruzada entre variáveis

```python
pd.crosstab(dados.Sexo, dados.Cor)
```

Criar tabelas legíveis

```python
sexo = {
    0: 'Masculino',
    1: 'Feminino'
}
```

```python
cor = {
    0: 'Indígena',
    2: 'Branca',
    4: 'Preta',
    6: 'Amarela',
    8: 'Parda'
}
```

Cruzamento profissional

```python
frequencia = pd.crosstab(
    dados.Sexo,
    dados.Cor
)

frequencia.rename(index=sexo, inplace=True)
frequencia.rename(columns=cor, inplace=True)

frequencia
```

```python
percentual = pd.crosstab(
    dados.Sexo,
    dados.Cor,
    normalize=True
) * 100
```

```python
renda_media = pd.crosstab(
    dados.Sexo,
    dados.Cor,
    values=dados.Renda,
    aggfunc='mean'
)
```

Heatmap

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.heatmap(renda_media, annot=True, fmt='.0f')

plt.title('Renda Média por Sexo e Cor')
plt.show()
```

Insights: Os dados mostram diferenças relevantes na renda média entre grupos demográficos, evidenciando desigualdades socioeconômicas entre sexo e cor.

Observa-se maior concentração de renda média entre homens brancos e amarelos, enquanto mulheres pretas e pardas apresentam menores médias salariais.

# Relação entre Média, Mediana e Moda

As medidas de tendência central ajudam a compreender o comportamento da distribuição dos dados.

# Medidas de Dispersão

## Desvio Médio Absoluto

As medidas de dispersão permitem avaliar o grau de variabilidade dos dados em relação à média.

```python
diferencas_absolutas = abs(
    dados['Renda'] - dados['Renda'].mean()
)

desvio_medio_absoluto = diferencas_absolutas.mean()

desvio_medio_absoluto
```

### Insight

O desvio médio absoluto evidencia a elevada dispersão da variável renda em relação à média, indicando forte desigualdade na distribuição.

A variância mede o grau de dispersão dos dados em relação à média.

```python
dados['Renda'].var()
```

```python
dados[['Renda', 'Idade', 'Altura']].var()
```

### Insight

A variável renda apresenta elevada variância, indicando grande dispersão dos valores em relação à média.


O desvio padrão mede a dispersão dos dados em relação à média, permitindo avaliar a variabilidade das observações.

```python
dados['Renda'].std()
```

```python
dados[['Renda', 'Idade', 'Altura']].std()
```

### Insight

A variável renda apresenta elevado desvio padrão, indicando alta variabilidade dos rendimentos na população analisada.

```python
print('Média:', dados['Renda'].mean())
print('Desvio padrão:', dados['Renda'].std())
```

A elevada dispersão da renda evidencia grande heterogeneidade econômica na distribuição dos dados.
